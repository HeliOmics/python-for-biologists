# Data Types

## Python's 4 Basic Data Types

### 1. String in Biology
### 2. Integers in Biology
### 3. Floats in Biology
### 4. Booleans in Biology


```python
# String in Biology
gene_name  = "BRCA1"
organism   = "Homo sapeins"
dna_seq    = "ATGCTTAGCATGCAT"

print(gene_name)
print(organism)
print(dna_seq)
```

```python
# Integers (whole numbers) in Biology
sequence_length = 1863     # base pairs
exon_count      = 8        # number of exons
chromosome      = 17       # chromosome number

print(sequence_length)
print(exon_count)
print(chromosome)


```

```python
# Floats (decimals numbers) in Biology


molecular_weight = 23.457  # kDa
gc_content       = 0.542   # as a proportion
melting_temp     = 68.5    # °C

print(molecular_weight)
print(gc_content)
print(melting_temp)
```

```python
# Booleans in Biology
# Boolean variables — answering yes/no questions about your data
is_coding_sequence = True
has_stop_codon     = True
is_reverse_strand  = False

print(f"Is coding: {is_coding_sequence}")
print(f"Stop codon: {has_stop_codon}")
print(f"Is reverse strand: {is_reverse_strand}")
```

```python
# Booleans in Biology
# A real gene record — BRCA1 exon 11 fragment
gene_name        = "BRCA1"
sequence_length  = 1863          # base pairs (int)
gc_content       = 0.421         # proportion (float)
is_coding        = True          # whether the gene is protein-coding or not. True is a Boolean value
is_complete      = sequence_length > 1000   # True — evaluated automatically!

print(f"Gene: {gene_name} | Coding: {is_coding} | Complete: {is_complete}")
```

# type() in Python

### The type() function tells you the data type or class of a value or variable

```python
gene_name = "TP53"
sequence_length = 42
gc_content = 2.87
nucleotides = ["adenine", "thymine", "guanine", "cytosine"]
codon_table = {1: "start", 2: "elongation", 3: "stop"}


print(type(gene_name))
print(type(sequence_length))
print(type(gc_content))
print(type(nucleotides))
print(type(codon_table))
```
