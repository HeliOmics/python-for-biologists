# Episode 2: Storing and Working with Basic DNA Data

## Store a DNA sequence, gene name, and chromosome number as variables.
## Check each variable's data type.
## Calculate the sequence length using len()
## Calculate GC content and store it as a float
## Print a formatted summary using all four variable types together


```python
# Store a DNA sequence, gene name, and chromosome number as variables
dna_sequence = "ATGCGCTAGCTAGGCTAGCATGCGATCGATCGATCGTAGC"
gene_name = "BRCA1"
chromosome_number = 17
```

```python
# Check each variable's data type.
print("---Data Types---")
print(f'dna_sequence is of type: {type(dna_sequence)}')
print(f'gene_name is of type:{type(gene_name)}')
print(f'chromosome_number is of type:{type(chromosome_number)}')
print()
```

```python
#Calculate the sequence length using len()
sequence_length= len(dna_sequence)
print(sequence_length)
```

```python
## Calculate GC content and store it as a float
g_count = dna_sequence.count("G")
print(g_count)
c_count = dna_sequence.count("C")
print(c_count)
gc_content = (g_count+c_count)/sequence_length
print(gc_content)
type(gc_content)
```

```python
#Print a formatted summary using all four variable types together
print("---Gene Summary---")
print(f'Gene:{gene_name} (Chromosome {chromosome_number})')
print(f'Sequence:{dna_sequence}')
print(f'Length:{sequence_length} bases')
print(f'GC Content:{gc_content:.2%}')
```
