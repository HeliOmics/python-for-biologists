# Booleans in Biology

### Booleans are the simplest type—only two possible values: True or False. They answer yes/no questions about your data.



```python
# Basic true/false facts about a gene or sample

is_mutated          = True       #sequence has mutation             
is_coding_region    = False      #sequenc contain a coding region
is_reference_strand = True       #sequence has reference strand 
sample_passed_qc    = False      #sample passed quality control

print(f"Is mutated: {is_mutated}")
print(f"Is coding region: {is_coding_region}")
print(f"Passed quality control: {sample_passed_qc}")
```

```python
# Booleans from comparisons (very common in real bioinformatics code)

gene_length       = 1200
is_long_gene      = gene_length > 1000        # True
print(is_long_gene)

read_quality_score = 28
is_high_quality    = read_quality_score >= 30   # False
print(is_high_quality)

chromosome_number  = 21
is_autosome        = chromosome_number not in [23, 24]   # True (23/24 often represent X/Y)
print(is_autosome)
```

```python
#Checking sequence content

dna_sequence         = "ATGCGCATG"

contains_start_codon = "ATG" in dna_sequence
print(contains_start_codon)   # True

is_valid_dna         = all(base in "ATGC" for base in dna_sequence)
print(is_valid_dna)   # True — checks every character is a valid DNA base
```

```python
# GC content threshold check
gc_content    = 0.62
is_gc_rich    = gc_content > 0.6
print(is_gc_rich)   # Tr
```

```python
# Using booleans in an if statement 
mutation_count   = 3

if mutation_count > 0:
    has_mutations = True
else:
    has_mutations = False

print(f"Has mutations: {has_mutations}")
```

```python
#Combining booleans with and / or
is_coding_region = True   #sequence has coding region
is_mutated       = True   #sequence has mutation

# Flag only if it's both a coding region AND mutated
needs_review     = is_coding_region and is_mutated
print(needs_review)   # True
```

# Simple biology-related example using Boolean operators (and, or, not)

####  and = both conditions must be true for and,
#### or = at least one condition must be true, 
#### not = reverse the condition


```python
# Example 1: Checking if a DNA sequence is valid (and)
sequence = "ATGCGAT"

if len(sequence) > 0 and "N" in sequence:
    print("DNA sequence is valid")
    
else:
    print("DNA sequence is invalid")
        
    
#Note: both conditions must be True because of and.
```

```python
# Example 2: Identifying a Purine (or)

base = "G"

if base =="G" or base =="A":
    print("Purine")
    
else:
    print("Pyrimidine")
    
#Note: Adenine (A) and Guanine (G) are purines. Only one condition needs to be true because of or.

```

```python
# Example 3: Rejecting Non-DNA Bases (not)

base = "A"

if not base=="U":
    print("It's a DNA base")
    
else:
    print("It's a RNA base")
    
#Note: Uracil (U) is found in RNA, not DNA and not reverse the condition
```
